# NIFTY 50 PROJECT REPORT
## Quantitative Analytics, Volatility Modeling & Portfolio Optimization on Nifty 50 (2000–2021)

---

### Executive Summary
This project analyzes more than two decades (January 2000 – April 2021) of historical equity market data from the National Stock Exchange of India (NSE) covering Nifty 50 constituent stocks. 

The study:
1. Identifies the empirical market indicators that drive equity volatility and price unpredictability.
2. Selects high-conviction, resilient assets to construct an optimized multi-asset portfolio that outperforms the benchmark.
3. Feature engineers technical and statistical indicators to train predictive machine learning models using chronological validation.
4. Deploys an interactive two-page Executive Power BI Dashboard delivering portfolio analytics and macro market cycle breakdowns.

---

### 1. Data Ingestion, Architecture & Standardization
- **Dataset Scale:** 235,192 daily market trading observations spanning 21.3 years across 52 unique tickers and 14 distinct economic sectors.
- **Data Quality:** Zero missing values across core pricing series (`Open`, `High`, `Low`, `Close`, `VWAP`, `Volume`).
- **Entity Normalization:** Legacy ticker changes were resolved and unified (e.g., `MUNDRAPORT` $\rightarrow$ `ADANIPORTS`, `UTIBANK` $\rightarrow$ `AXISBANK`, `BAJAUTOFIN` $\rightarrow$ `BAJFINANCE`).
- **Sector Mapping:** Left-joined metadata to associate each equity with its primary industry sector.

---

### 2. Task 1: Volatility & Price Unpredictability Analysis (Last Decade: 2011–2021)
Task 1 evaluates the statistical relationships between trading indicators and stock volatility over the 2011–2021 period:
- **Intraday High-Low Range Spread ($r = +0.3114$):** The primary contemporaneous signal of elevated volatility regimes.
- **Trade Density & Absolute Returns ($r = +0.3031$, $r = +0.2505$):** Surges in total transactions accompany panic selling and volatile rebalancing.
- **Deliverable Volume Percentage ($r = -0.2514$):** A statistically significant negative correlation. When delivery percentages are high ($>60\%$), institutional long-term holding dampens volatility. When delivery percentages fall, intraday speculation drives unpredictable price swings.

### 4. Task 4: Feature Engineering & Machine Learning
- **Target Formulation:** Next-day binary directional movement ($Y_{t+1}$) and 5-day weekly momentum direction ($Y_{t+5}$).
- **Engineered Predictors:** Lags (`Return_1D`, `Return_2D`, `Return_5D`), 14-Day RSI, Moving Average Ratio (`SMA_10 / SMA_50`), Intraday Spread, and Volume Surge.
- **Validation Scheme:** Chronological time-series split (Train: 2011–2018 | Test: 2019–2021). No data leakage or look-ahead bias.
- **Modeling Results:**
  - **1-Day Horizon:** 46.53% accuracy, consistent with the Random Walk Theory on high-frequency noise.
  - **5-Day Horizon:** **54.72% accuracy** (Precision: 55%, Recall: 93% for upward regimes).
  - **Feature Importances:** Intermediate momentum (`Return_5D`: 0.181) and trend alignment (`SMA_Ratio`: 0.146) were the leading predictive signals.

### 5. Task 3: Interactive Executive Dashboard Architecture
Built in Power BI Desktop across two dedicated reporting interfaces:
1. **Page 1 — "Last Decade Analysis (2011–2021)":** Focuses on Task 1 and Task 2. Features portfolio wealth trajectories, Sharpe ranking tables, and interactive stock slicers.
2. **Page 2 — "2 Decades Portfolio (2000–2021)":** Addresses the broad 20-year problem statement. Tracks macro performance through historical events (the 2000 Dot-Com crash, 2008 GFC, and 2020 COVID shock), sector CAGR rankings, and a 20-year risk-return scatter matrix.

---

### 6. Conclusions & Strategic Takeaways
1. **Speculation vs. Delivery:** Low delivery volumes indicate speculative trading environments prone to sharp drawdowns.
2. **Sector Allocation Over Timing:** Non-cyclical consumer staples and private financial institutions delivered consistent multi-decade compound growth, whereas cyclical commodities experienced prolonged drawdown regimes.
3. **Multi-Asset Diversification:** Pairing high-growth assets with defensive consumer anchors preserved capital during major market crises, delivering consistent risk-adjusted alpha over the benchmark index.
